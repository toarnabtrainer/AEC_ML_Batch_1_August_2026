# **Logistic Regression**

### What Is Logistic Regression?
Logistic regression is a supervised learning algorithm used to predict a dependent categorical target variable. In essence, if you have a large set of data that you want to categorize, logistic regression may be able to help. 

### What Are the Different Types of Logistic Regression?
There are three main types of logistic regression: binary, multinomial and ordinal. They differ in execution and theory. Binary regression deals with two possible classes, essentially: yes or no. Multinomial logistic regression deals with three or more classes. And ordinal logistic regression deals with three or more classes in a predetermined order.

### Binary Logistic Regression
Binary logistic regression in the case of classifying an object as an animal or not an animal—it’s an either/or solution. There are just two possible outcome answers. This concept is typically represented as a 0 or a 1 in coding. Examples include:

* Whether or not to lend to a bank customer (outcomes are yes or no).
* Assessing cancer risk (outcomes are high or low).
* Will a team win tomorrow’s game (outcomes are yes or no).

### Multinomial Logistic Regression
Multinomial logistic regression is a model where there are multiple classes that an item can be classified as. There is a set of three or more predefined classes set up prior to running the model. Examples include:

* Classifying texts into what language they come from.
* Predicting whether a student will go to college, trade school or into the workforce.
* Does your cat prefer wet food, dry food or human food?

### Ordinal Logistic Regression
Ordinal logistic regression is also a model where there are multiple classes that an item can be classified as; however, in this case an ordering of classes is required. Classes do not need to be proportionate. The distance between each class can vary. Examples include:

* Ranking restaurants on a scale of 0 to 5 stars.
* Predicting the podium results of an Olympic event.
* Assessing a choice of candidates, specifically in places that institute ranked-choice voting.

### **Why Harmonic Mean not Arithmetic Mean?**
### The Problem with the Arithmetic Mean

Let's imagine a scenario where we have a classification model. Precision and recall are two important metrics for evaluating it.

* **Precision** tells you how many of the positive predictions were actually correct ($TP / (TP + FP)$). It focuses on the quality of the positive predictions.
* **Recall** tells you how many of the actual positive cases were correctly identified ($TP / (TP + FN)$). It focuses on the model's ability to find all the positive cases.

Now, consider a highly imbalanced dataset, for example, a fraud detection system where only 1% of transactions are fraudulent.

* **Model A** is very conservative and only predicts a transaction as fraudulent if it's 99.9% sure. This model might have very few false positives, so its **precision is very high (e.g., 0.95)**. However, it misses many fraudulent transactions, so its **recall is very low (e.g., 0.10)**.
* **Model B** is very aggressive and flags many transactions as fraudulent, even if it's not very sure. This model might catch almost all fraudulent transactions, so its **recall is very high (e.g., 0.99)**. However, it also flags a lot of legitimate transactions as fraudulent, so its **precision is very low (e.g., 0.05)**.

If we were to use the **arithmetic mean** to combine precision and recall:

* **Arithmetic Mean (Model A):** $(0.95 + 0.10) / 2 = 0.525$
* **Arithmetic Mean (Model B):** $(0.05 + 0.99) / 2 = 0.52$

As you can see, both models get a similar score. This is misleading! Model A has high precision but poor recall, while Model B has poor precision but high recall. Neither is a particularly good model, but the arithmetic mean makes them look average. It doesn't penalize the model for having a low value in one of the metrics.

### The Power of the Harmonic Mean

The harmonic mean is different. It's a special type of average that is used for rates and ratios. Its formula is:

$$F1 = 2 \cdot \frac{\text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}$$

Here's why it's a much better choice for the F1 score:

1.  **It severely penalizes models with a low value in either precision or recall.** For the F1 score to be high, both precision and recall must be high. If one of them is very low, the F1 score will also be very low. Let's revisit our examples:
    * **F1 Score (Model A):** $2 \cdot (0.95 \cdot 0.10) / (0.95 + 0.10) \approx 0.17$
    * **F1 Score (Model B):** $2 \cdot (0.05 \cdot 0.99) / (0.05 + 0.99) \approx 0.09$

    This gives a much more accurate picture. The F1 scores for both models are low, indicating that they are not well-balanced and therefore not very good classifiers. This is what we would expect.

2.  **It gives more weight to the smaller value.** The harmonic mean is always less than or equal to the arithmetic mean, and it's heavily influenced by the smallest number in the set. This behavior is exactly what we want when balancing precision and recall. A model that has a huge disparity between precision and recall should be considered worse than a model where they are more balanced, even if their arithmetic mean is the same.

3.  **It's a "rate-of-rates" average.** Precision and recall are ratios, not simple numbers. The harmonic mean is the correct average to use when dealing with rates.

In essence, the F1 score using the harmonic mean serves as a robust metric for models, especially when dealing with imbalanced datasets. It forces a model to perform well on *both* precision and recall to achieve a high score, providing a single, trustworthy number that represents the model's overall effectiveness.
