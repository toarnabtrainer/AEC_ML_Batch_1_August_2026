## **Reshape**

```python
matrix = np.arange(20); print(matrix)
matrix = np.arange(20).astype(float).reshape((4, 5)); print(matrix)
matrix = np.arange(20).reshape((2, -1, 5)); print(matrix)
matrix = np.arange(20).reshape((2, 2, -1)); print(matrix)
matrix = np.arange(20).reshape((2, 2, 5)); print(matrix)
```

```python
matrix = np.arange(10, dtype = float).reshape((2, 5))
print(matrix)
print("Stored in C-style order (Row major):")
cmatrix = matrix.copy(order = 'c')
print(cmatrix)
for data in np.nditer(cmatrix):
    print(data, end = ", ")
```

```python
matrix = np.arange(10, dtype = float).reshape((2, 5))
print(matrix)
print("Stored in F-style order (Column major):")
fmatrix = matrix.copy(order = 'f')
print(fmatrix)
for data in np.nditer(fmatrix):
    print(data, end = ", ")
print("\n", fmatrix.shape, fmatrix.size, len(fmatrix))
```

## **Append, Insert, Delete, and Sort**

```python
matrix = np.arange(10, dtype = np.int8); print(matrix, matrix.shape)
matrix = np.append(matrix, [10, 11, 12]); print(matrix, matrix.shape)
matrix = np.insert(matrix, 3, [111, 222, 444]); print(matrix, matrix.shape)
matrix = np.delete(matrix, [3, 4, 7]); print(matrix, matrix.shape)
matrix = np.sort(matrix); print(matrix, matrix.shape) # by default ascending
matrix = -np.sort(-matrix); print(matrix, matrix.shape) # descending order
```

## <b>Concatenation and Stack of Two Arrays</b>

```python
arr1 = np.array([[11, 22, 33], [44, 55, 66]]); print(arr1, arr1.shape)
arr2 = np.array([[10, 20, 30], [40, 50, 60]]); print(arr2, arr2.shape)
# both two arrays are of same dimensions
print("Concatenating two arrays along axis-0:")
result = np.concatenate((arr1, arr2), axis = 0); print(result, result.shape)
print("Concatenating two arrays along axis-1:")
result = np.concatenate((arr1, arr2), axis=1); print(result, result.shape)
```

```python
print(arr1, arr1.shape)
print(arr2, arr2.shape)
result = np.stack((arr1, arr2), axis = 0); print(result, result.shape)
result = np.stack((arr1, arr2), axis = 1); print(result, result.shape)
```

## **Using Boolean Arrays**

```python
arr1 = np.array([100, -21, 31, 150, -40, 601, -700])
print(arr1, arr1.shape)
print(arr1 >= 0, arr1[arr1 >= 0])
print(arr1 < 0, arr1[arr1 < 0])
print(arr1 >= 0, ~(arr1 >= 0))
print(arr1[~(arr1 >= 0)])
print(arr1 % 2 == 0, arr1[arr1 % 2 == 0], arr1[~(arr1 % 2 == 0)])
print(arr1[(arr1 % 2 == 0) & (arr1 % 3 == 0)])
print(arr1[(arr1 % 2 == 0) | (arr1 % 3 == 0)])
```

```python
arr1 = np.array(["Bob", "Tom", "Jac", "Bob", "Pam"])
print(arr1, arr1.shape)
print(arr1[arr1 == 'Bob'], arr1[arr1 == 'bob'])
print(arr1[~(arr1 == "Bob")], arr1[arr1 != 'Bob'])
arr1[(arr1 == 'Bob')] = "Pat"
print(arr1, arr1.shape)
print(np.unique(arr1))
# print(np.unique(arr1).sort())
arr2 = np.unique(arr1)
arr2.sort()
print(arr2)
```

```python
# calculating Eucledian distance between two points
# pt1(x1, y1) and pt2(x2, y2), then dist = ((x1 - x2)^2 + (y1 - y2)^2)^0.5
pt1 = np.random.randn(10)  # point in a n-dimensional space
pt2 = np.random.randn(10)
print(pt1, "\n", pt2)
e_dist = 0
for i in range(len(pt1)):
    e_dist += (pt1[i] - pt2[i]) ** 2
print("Eucledian distance =", np.sqrt(e_dist))
print("Eucledian distance =", np.sqrt(np.sum((pt1 - pt2) ** 2)))
print("Eucledian distance =", np.sqrt(np.sum(np.subtract(pt1, pt2) ** 2)))
print("Eucledian distance =", np.sqrt(np.sum(np.pow(np.subtract(pt1, pt2), 2))))
```

## **Broadcasting**

```python
a1 = np.array([[ 0, 0, 0],
               [10,10,10],
               [20,20,20],
               [30,30,30]])
a2 = np.array([[0], [10], [20], [30]])
b1 = np.array([[ 0, 1, 2],
               [ 0, 1, 2],
               [ 0, 1, 2],
               [ 0, 1, 2]])
b2 = np.array([ 0, 1, 2])
print(a1 + b1)
print(a1 + b2)
print(a2 + b1)
print(a2 + b2)
```

```python
print(np.add(a1, b1))
print(np.add(a1, b2))
print(np.add(a2, b1))
print(np.add(a2, b2))
```
