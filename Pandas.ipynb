# **Introduction to Pandas**
* **It is often said that 80% of data analysis activity is spent on the data cleaning and preparation of our data. To get the grip over this activity, this section will focus on a small but important aspect of data manipulation and cleaning with Pandas.**

## **Inbuilt Data Structure in Pandas**
**There are two types of data structures are there in Pandas -** <br>

* **Series -** It is one-dimensional labeled array capable of holding any data type (integer, string, float, Python objects etc.) of data. The axis is collectively referred to as index. <br>

* **Data Frame -** It is a two-dimensional labeled data structure with columns of potentially different types. You can think of it like a spreadsheet or SQL Table or a Series of objects <br>

## **Series Data Structure**
**pandas.core.series.Series(data, index, dtype, copy)** <br>
* **data -** data takes various forms like ndarray, list, constants, dictionary etc. <br>
* **index -** it is unique and hashable for easy identification. <br>
* **dtype -** it is for data type. <br>
* **copy -** copy data, and its default value is False. It only affects for Series or one dimensional ndarray inputs. <br>

```python
# importing required modules
import numpy as np
import pandas as pd
print(np.__version__)  # magic variables
print(pd.__version__)
```

```python
# creating empty series
s = pd.Series()
print(s)
print(type(s), len(s), id(s))
```

```python
# creating series from ndarray
arr_data = np.array(['apple', 'pineapple', 'lime', 'lemon'])
s = pd.Series(data = arr_data)
print(s)
print(type(s), len(s), id(s))
print(s[0], s[3])
```

```python
arr_data = np.array([100, 300, 200, 500])
print(arr_data)
s = pd.Series(data = arr_data, copy = False)
print(s)
arr_data[0] = 111
s[3] = 555
print(arr_data)
print(s)
```

```python
arr_data = np.array([100, 300, 200, 500])
print(arr_data)
s = pd.Series(data = arr_data, copy = True)
print(s)
arr_data[0] = 111
s[3] = 555
print(arr_data)
print(s)
```

```python
arr_data = np.array(['apple', 'pineapple', 'lime', 'lemon'])
arr_index = [100, 200, 400, 300]
s = pd.Series(data = arr_data, index = arr_index)
print(s)
print(s[100], s[400])
```

```python
arr_data = np.array(['apple', 'pineapple', 'lime', 'lemon'])
arr_index = [100, 200, 100, 300]
s = pd.Series(data = arr_data, index = arr_index)
print(s)
print(s[100])
print(s[200])
```

## **Data Frame Data Structure**

```python
# create dataframe from list of lists
user_data = [['alice', 19, 'F', 'student'], ['john', 26, 'M', 'student']]
user_column = ['name', 'age', 'gender', 'job']
user1 = pd.DataFrame(data = user_data, columns = user_column)
display(user1)
```

```python
# create dataframe from a dictionary
user_data = dict(name=['eric','paul'], age=[22,58], gender=['M','F'], job=['student','manager'])
print(user_data)
user2 = pd.DataFrame(data = user_data)
display(user2)
```

```python
# create dataframe from a dictionary
user_data = {'name':['peter','julie'], 'age':[33,44], 'gender':['M','F'], 'job':['engineer','scientist']}
user3 = pd.DataFrame(data = user_data)
display(user3)
```

```python
users = pd.concat([user1, user2, user3])
display(users)
```

```python
users = pd.concat([user1, user2, user3], ignore_index=True)
display(users)
```

```python
# convert dataframe to numpy ndarray
arr_data = users.to_numpy()
print(arr_data)
arr_data = np.array(users)
print(arr_data)
```

## **Join Operations on DataFrames**

```python
dict_data = dict(name=['alice','john','eric','julie','smith'], height=[165,180,175,169,180])
print(dict_data)
user4 = pd.DataFrame(data = dict_data)
display(user4)
```

```python
# inner join: all common rows from both data frames
merge_inner = pd.merge(users, user4, on = "name", how = "inner")
display(merge_inner)
```

```python
# inner join: all common rows from both data frames
merge_inner = pd.merge(users, user4, on = "name")
display(merge_inner)
```

```python
# left outer join: all from left and matching from right
merge_left = pd.merge(users, user4, on = "name", how = "left")
display(merge_left)
```

```python
# right outer join: matching from left and all from right
merge_right = pd.merge(users, user4, on = "name", how = "right")
display(merge_right)
```

```python
# full outer join: all from left and right matching and not matching
merge_full = pd.merge(users, user4, on = "name", how = "outer")
display(merge_full)
```
